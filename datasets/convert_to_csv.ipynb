```python
# cell 0 - convert .xlsx to .csv (supports single or all sheets)
from pathlib import Path
import pandas as pd

def convert_xlsx_to_csv(xlsx_path, output_dir=None, sheet_name=None, index=False, encoding='utf-8'):
    """
    Convert an .xlsx file to one or more .csv files.
    - xlsx_path: path to the .xlsx file (str or Path)
    - output_dir: directory to write csv(s). If None, use xlsx file's parent.
    - sheet_name: None (all sheets), sheet name, or sheet index. Matches pandas.read_excel argument.
    - index: whether to write row index to csv (default False).
    - encoding: csv file encoding (default 'utf-8').
    Returns list of written csv file paths.
    """
    xlsx_path = Path(xlsx_path)
    if output_dir is None:
        output_dir = xlsx_path.parent
    output_dir = Path(output_dir)
    output_dir.mkdir(parents=True, exist_ok=True)

    # Read Excel; sheet_name=None returns dict of DataFrames
    data = pd.read_excel(xlsx_path, sheet_name=sheet_name, engine='openpyxl')

    written_files = []
    if isinstance(data, dict):
        # multiple sheets
        for sheet, df in data.items():
            safe_sheet = "".join(c if c.isalnum() or c in " _-." else "_" for c in str(sheet)).strip()
            out_name = f"{xlsx_path.stem}_{safe_sheet}.csv"
            out_path = output_dir / out_name
            df.to_csv(out_path, index=index, encoding=encoding)
            written_files.append(out_path)
    else:
        # single sheet (DataFrame)
        out_name = f"{xlsx_path.stem}.csv"
        out_path = output_dir / out_name
        data.to_csv(out_path, index=index, encoding=encoding)
        written_files.append(out_path)

    return written_files

# Example usage:
# files = convert_xlsx_to_csv("data.xlsx")                      # all sheets -> multiple CSVs
files = convert_xlsx_to_csv("Microsource_Pharmacon1600_M00966-70.xlsx", sheet_name=0)        # only first sheet
# files = convert_xlsx_to_csv("data.xlsx", output_dir="csvs")   # write to csvs/ directory
# print(files)
```
